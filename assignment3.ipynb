With Colab you can harness the full power of popular Python libraries to analyze and visualize data. The code cell below uses **numpy** to generate some random data, and uses **matplotlib** to visualize it. To edit the code, just click the cell and start editing.

```
import numpy as np
import IPython.display as display
from matplotlib import pyplot as plt
import io
import base64

ys = 200 + np.random.randn(100)
x = [x for x in range(len(ys))]

fig = plt.figure(figsize=(4, 3), facecolor='w')
plt.plot(x, ys, '-')
plt.fill_between(x, ys, 195, where=(ys > 195), facecolor='g', alpha=0.6)
plt.title("Sample Visualization", fontsize=10)

data = io.BytesIO()
plt.savefig(data)
image = F"data:image/png;base64,{base64.b64encode(data.getvalue()).decode()}"
alt = "Sample Visualization"
display.display(display.Markdown(F"""![{alt}]({image})"""))
plt.close(fig)
```

```
import pandas as pd

purchase = pd.read_csv("purchase_data.csv")
customer = pd.read_csv("customer_data.csv")

print(purchase.head())
print(customer.head())
```

```
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

```
purchase['Revenue'] = (purchase['price'] * purchase['quantity']) - purchase['discount'] + purchase['tax']
```

```
top_products = purchase.groupby('product_name')['Revenue'].sum().sort_values(ascending=False).head(5)
print(top_products)
```

Objective:
The purpose of this analysis is to identify the products that contribute the most to total revenue.

Explanation:
Revenue was calculated using the formula:
Revenue = (Price × Quantity) − Discount + Tax

After calculating revenue for each transaction, the total revenue for each product was aggregated. The top 5 products contributing the highest revenue were identified.

Interpretation:
The bar chart shows that a small number of products generate a significant portion of total revenue. These products are the key drivers of business performance.

Conclusion:
The company should focus on these high-performing products by optimizing pricing strategies, reducing unnecessary discounts, and ensuring their availability.

```
plt.figure()
plt.bar(top_products.index, top_products.values)
plt.title("Top 5 Products by Revenue")
plt.xlabel("Product Name")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

```
purchase['order_date'] = pd.to_datetime(purchase['order_date'])
purchase['shipping_date'] = pd.to_datetime(purchase['shipping_date'])
```

```
purchase['Shipping_Days'] = (purchase['shipping_date'] - purchase['order_date']).dt.days
```

```
within_3 = purchase[purchase['Shipping_Days'] <= 3]['shipping_cost'].mean()
after_3 = purchase[purchase['Shipping_Days'] > 3]['shipping_cost'].mean()

print("≤ 3 days:", within_3)
print("> 3 days:", after_3)
```

Objective:
To analyze whether shipping time affects shipping cost.

Explanation:
Orders were divided into two groups:

Orders shipped within 3 days
Orders shipped after 3 days

The average shipping cost for both groups was calculated.

Interpretation:
If the shipping cost is higher for delayed orders, it indicates inefficiencies in logistics.

Conclusion:
The company should optimize its shipping process to ensure faster deliveries and reduce costs, improving both efficiency and customer satisfaction.

```
merged = pd.merge(purchase, customer, on='customer_id')
```

```
male_income = merged[merged['gender'] == 'Male']['income']
female_income = merged[merged['gender'] == 'Female']['income']

t_stat, p_value = stats.ttest_ind(male_income, female_income)

print("P-value:", p_value)
```

Objective:
To determine whether there is a significant difference in income between male and female customers.

Hypotheses:

H0 (Null Hypothesis): There is no significant difference in average income between genders
H1 (Alternative Hypothesis): There is a significant difference

Explanation:
An independent t-test was performed to compare the average income of male and female customers.

Interpretation:

If p-value < 0.05 → Reject H0 → Significant difference exists
If p-value > 0.05 → Accept H0 → No significant difference

Conclusion:
Based on the result, the company can decide whether gender-based targeting strat

```
merged['Product_Category'] = merged['product_name'].apply(lambda x: x.split()[0])

import pandas as pd
contingency = pd.crosstab(merged['Product_Category'], merged['gender'])

chi2, p, dof, expected = stats.chi2_contingency(contingency)

print("P-value:", p)
```

Objective:
To determine whether there is an association between product categories and customer gender.

Explanation:
A contingency table was created, and a Chi-square test was applied.

Interpretation:

If p-value < 0.05 → Significant association exists.

If p-value > 0.05 → No association

Conclusion:
If an association exists, the company can design gender-specific product marketing strategies.
